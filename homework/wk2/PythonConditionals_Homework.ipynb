```python
#Write a function that converts a numeric grade to a letter grade:

#90-100: A, 80-89: B, 70-79: C, 60-69: D, Below 60: F
def convert_to_letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

# Test Task 3
print("Task 3 Test:")
print(f"99 points is a {convert_to_letter_grade(99)} grade")
print(f"83 points is a {convert_to_letter_grade(83)} grade")
print(f"77 points is a {convert_to_letter_grade(77)} grade")
print(f"61 points is a {convert_to_letter_grade(61)} grade")
print(f"12 points is a {convert_to_letter_grade(12)} grade")
```

```python
#Write a function to determine user access based on:

#Login status (True or False)
#Admin status (True or False)
#Premium membership (True or False)

#Conditions:

#If not logged in, print "Please log in to access the site".
#If admin, print "Welcome admin! Full access granted".
#If premium user, print "Welcome premium user! Premium content access granted".
#Otherwise, print "Welcome! Basic access granted".

def check_access(is_logged_in, is_admin, is_premium):
    # Outer conditional: Check if the user is even logged in first
    if is_logged_in:
        # Inner conditional: Now that we know they are logged in, check roles
        if is_admin:
            print("Welcome admin! Full access granted")
        elif is_premium:
            print("Welcome premium user! Premium content access granted")
        else:
            print("Welcome! Basic access granted")
    else:
        # This belongs to the outer "if", triggered if is_logged_in is False
        print("Please log in to access the site")

# Test Task 4
print("Task 4 Test:")
check_access(False, False, False)  # Output: Please log in to access the site
check_access(True, True, False)    # Output: Welcome admin! Full access granted
check_access(True, False, True)    # Output: Welcome premium user! Premium content access granted
check_access(True, False, False)   # Output: Welcome! Basic access granted
print()
```

```python
#Write a function that returns "Even" or "Odd" based on whether a number is divisible by 2, using the ternary operator.
def check_even_odd(number):
    # Syntax: value_if_true if condition else value_if_false
    return "Even" if number % 2 == 0 else "Odd"

# Test Task 5
print("Task 5 Test:")
print(f"11 is {check_even_odd(11)}") # Output: 5 is Odd
print(f"32 is {check_even_odd(32)}") # Output: 8 is Even
```

```python
#Write a function to check if a year is a leap year.

#A leap year is divisible by 4, except when divisible by 100 unless also divisible by 400.

def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False

# Test Task 6
print("Task 6 Test:")
print(f"2020 is a leap year: {is_leap_year(2020)}")  # Output: True
print(f"2000 is a leap year: {is_leap_year(2000)}")  # Output: True
print(f"1900 is a leap year: {is_leap_year(1900)}")  # Output: False
print(f"2022 is a leap year: {is_leap_year(2022)}")  # Output: False
print(f"1976 is a leap year: {is_leap_year(1976)}")  # Output: True
```

```python
#Refactor the function below to be more readable and efficient:

def classify_item(item_type, price, in_stock):
    tier = "Affordable" if price < 100 else "Premium"
    category = "electronics" if item_type == "electronics" else "non-electronics"
    availability = "in stock" if in_stock else "out of stock"
    return f"{tier} {category} {availability}"

print(classify_item("electronics", 50, True))    # Affordable electronics in stock
print(classify_item("electronics", 150, False))  # Premium electronics out of stock
print(classify_item("clothing", 50, False))      # Affordable non-electronics out of stock
print(classify_item("clothing", 150, True))      # Premium non-electronics in stock
```
